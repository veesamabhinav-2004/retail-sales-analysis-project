```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("train.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
```

```python
sales_category = df.groupby('Category')['Sales'].sum()

sales_category.plot(kind='bar', title='Total Sales by Category')

plt.show()
```

```python
profit_category = df.groupby('Category')['Profit'].sum()

profit_category.plot(kind='bar', color='green', title='Profit by Category')

plt.show()
```

```python
sns.barplot(data=df, x='Region', y='Sales')

plt.title("Sales by Region")

plt.show()
```

```python
sales_trend = df.groupby('Year')['Sales'].sum()

sales_trend.plot(kind='line', marker='o', title='Sales Trend Over Years')

plt.xlabel("Year")
plt.ylabel("Total Sales")

plt.show()
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

top_products.plot(kind='bar', title='Top 10 Products by Sales')

plt.show()
```
